```python
from Montreal_UHI_toolbox import stations_mtl
import plotly.graph_objects as go
```

```python
obs = stations_mtl
```

```python
tasmin = obs.sel(time=slice('1980','2025'))['tasmin'].groupby('time.season')['JJA'].mean('station').groupby('time.year').mean('time')
# tasmax = obs.sel(time=slice('1980','2025'))['tasmax'].groupby('time.season')['JJA'].mean('station').groupby('time.year').mean('time')
# tas = obs.sel(time=slice('1980','2025'))['tas'].groupby('time.season')['JJA'].mean('station').groupby('time.year').mean('time')
```

```python
tasmin_clim = tasmin.sel(year=slice('1980','2010')).mean('year')
# tasmax_clim = tasmax.sel(year=slice('1980','2010')).mean('year')
# tas_clim = tas.sel(year=slice('1980','2010')).mean('year')
```

```python
tasmin_anom = tasmin - tasmin_clim
# tasmax_anom = tasmax - tasmax_clim
# tas_anom = tas - tas_clim
```

```python
years = tasmin.year.values

hover_text = [
    f"Time: {y} (JJA)<br>Temperature: {t:.2f}°C<br>Anomaly: {'+' if a > 0 else ''}{a:.2f}°C"
    for y, t, a in zip(years, tasmin, tasmin_anom)
]

fig = go.Figure(
    data=[
        go.Bar(
            name='Daily Minimum',
            x=years,
            y=tasmin_anom,
            marker_color='black',
            hovertext=hover_text,
            hoverinfo='text'
        )
    ]
)

fig.add_vrect(
    x0=1979.6, x1=2010.4,                  
    fillcolor='lightgrey', opacity=0.35,
    line_width=0
)

fig.add_annotation(
    x=(1980+2010)/2,              
    y=0.9,                
    text=f'1980-2010 Climatology ({tasmin_clim.values:.2f}°C)',  
    showarrow=False,
    arrowhead=2,
    ax=0,               
    ay=40,
    yref='paper'
)

fig.update_layout(title= dict(text='Montréal Region Observed Summer Daily Minimum Temperatures Compared to 1980-2010 Average',
                             xanchor='center',
                             x=0.5),
                  xaxis_title='Year', 
                  yaxis_title='T<sub>min</sub> Anomaly (°C)',
                  barmode='group')

fig.show()
# fig.write_html('/home/gulley/UHI_HW_MTL/info/plots/annual_climatology_tasmin.html')
```
