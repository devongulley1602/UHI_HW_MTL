```python
from Montreal_UHI_toolbox import get_outputs
```

```python
# Import summertime tas for both models
tas_JJA_C = get_outputs('tas',canopy='C').groupby('time.season')['JJA']
tas_JJA_T = get_outputs('tas',canopy='T').groupby('time.season')['JJA']

# Find the average temperature by the hour of day for all 23 summers
tas_JJA_diurnal_C = tas_JJA_C.groupby('time.hour').mean(dim='time')
tas_JJA_diurnal_T = tas_JJA_T.groupby('time.hour').mean(dim='time')
```

```python
# Save the gridded datasets
tas_JJA_diurnal_C.to_zarr('/runoff/gulley/St_Laurent/intermediates/sim/diurnal/tas_JJA_diurnal_C.zarr')
```

```python
tas_JJA_diurnal_T.to_zarr('/runoff/gulley/St_Laurent/intermediates/sim/diurnal/tas_JJA_diurnal_T.zarr')
```

```python
# Repeat this for the remaining subdaily fields
fields = ['hrss','hdx','hfls','hfss']
models = ['C','T']

for model in models:
    for field in fields:

        # Location to save
        save_dir = f'/runoff/gulley/St_Laurent/intermediates/sim/diurnal/{field}_JJA_diurnal_{model}.zarr'

        # Load, average by hour of day, save to save_dir
        diurnal = get_outputs(field,canopy=model).groupby('time.season')['JJA'].groupby('time.hour').mean(dim='time')
        diurnal.to_zarr(save_dir)

        print(f'Solved and saved for {field}_{model}')
```
