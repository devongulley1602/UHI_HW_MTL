```python
# Select for stations records only within the time frame of the simulation
# Monthly-averaged series by station type
urban_S = urban_stations.sel(time=slice('2000-01-01','2023-01-01')).groupby('time.month').mean(dim='time').mean(dim='station')
subur_S = subur_stations.sel(time=slice('2000-01-01','2023-01-01')).groupby('time.month').mean(dim='time').mean(dim='station')
rural_S = rural_stations.sel(time=slice('2000-01-01','2023-01-01')).groupby('time.month').mean(dim='time').mean(dim='station')

for field in ['tas','tasmax','tasmin']:
    monthly[field]['urban_S'] = urban_S[field]
    monthly[field]['rural_S'] = rural_S[field]
```

```python
total_fig = make_subplots(
    rows=1, cols=3,
    shared_yaxes=True,
    shared_xaxes=False,
    subplot_titles=[
        'Temperature', 'Maximum Daily Temperature', 'Minimum Daily Temperature'],
    vertical_spacing=0.12,  # reduce gap between rows
    horizontal_spacing=0.05
)

for field, col in zip(['tas', 'tasmax', 'tasmin'],[1, 2, 3]):

    # Adding simulation data
    for sim, urban_level, sim_name, colour, dash_style in zip(
        ['urban_C', 'rural_C', 'urban_T', 'rural_T'],
        ['urban', 'reduced urban', 'urban', 'reduced urban'],
        ['CLASS', 'CLASS', 'TEB+CLASS', 'TEB+CLASS'],
        ['blue', 'blue', 'red', 'red'],
        ['solid', 'dot', 'solid', 'dot']
    ):
        total_fig.add_trace(go.Scatter(
            x=months,
            y=monthly[field][sim] - 273.15*np.ones(12),
            name=f'{sim_name} ({urban_level})',
            legendgroup=f'{sim_name} {urban_level}',
            line=dict(color=colour, dash=dash_style,width=1.5),
            showlegend=(col == 1)
        ), row=1, col=col)

     # Adding station data
    for station_set, stationset_title, dash_style in zip(
        [urban_S, rural_S],
        ['urban', 'reduced urban'],
        ['solid', 'dot']
    ):
        total_fig.add_trace(go.Scatter(
            x=months,
            y=station_set[field],
            name=f'Observation ({stationset_title})',
            legendgroup=f'stations {stationset_title}',
            line=dict(color='black', dash=dash_style, width=1.5),
            showlegend=(col == 1)
        ), row=1, col=col)

    # Label x-axis
    total_fig.update_xaxes(
        title_text='Month',
        tickmode='array',
        tickvals=month_nums,
        ticktext=months,
        row=1, col=col
    )

# Y-axis on first subplot
dtick = 2
total_fig.update_yaxes(title_text="Temperature Cycles (°C)", row=1, col=1, dtick=dtick)
for col in [2, 3]:
    total_fig.update_yaxes(
        dtick=dtick,
        row=1, col=col
    )
# Overall layout
total_fig.update_layout(
    # height=450,
    # width=1300,
    template='plotly_white',
    legend=dict(
        # orientation='h',
        # yanchor='bottom',
        # y=-0.25,
        # xanchor='left',
        # x=0.5
    ),
    # margin=dict(t=60, b=60)
)

# total_fig.show()
total_fig.write_html('/home/gulley/UHI_HW_MTL/info/plots/monthly_grouped_temperature_cycles.html')
# pio.write_image(total_fig,'/home/gulley/UHI_HW_MTL/info/plots/monthly_grouped_temperature_cycles.pdf', format='pdf')
```

```python
total_fig = make_subplots(
    rows=1, cols=3,
    shared_yaxes=True,
    shared_xaxes=False,
    subplot_titles=[
        'Temperature', 'Maximum Daily Temperature', 'Minimum Daily Temperature'],
    vertical_spacing=0.12,
    horizontal_spacing=0.05
)

for field, col in zip(['tas', 'tasmax', 'tasmin'],[1, 2, 3]):
    
    # UHI effect
    for urban_data, rural_data,colour,sim_name in zip(['urban_C','urban_T','urban_S'],
                                      ['rural_C','rural_T','rural_S'],
                                      ['blue','red','black'],
                                      ['CLASS','TEB+CLASS','Observation']):
        total_fig.add_trace(go.Scatter(
            x=months,
            y=np.array(monthly[field][urban_data]) - np.array(monthly[field][rural_data]),
            name=f'{sim_name}',
            legendgroup=f'{sim_name} UHI',
            line=dict(color=colour, dash='solid',width=1.5),
            showlegend=(col == 1)
        ), row=1, col=col)

dtick = 0.2
total_fig.update_yaxes(title_text="UHI Temperature Effect (°C)", row=1, col=1,dtick=dtick)
for col in [2, 3]:
    total_fig.update_yaxes(
        dtick=dtick,
        row=1, col=col
    )
# Overall layout
total_fig.update_layout(
    # height=450,
    # width=1300,
    template='plotly_white',
    legend=dict(
    ),
    # margin=dict(t=60, b=60)
)

total_fig.show()
# total_fig.write_html('/home/gulley/UHI_HW_MTL/info/plots/monthly_grouped_UHI_temperature_cycles.html')
```

```python
total_fig = make_subplots(
    rows=1, cols=3,
    shared_yaxes=True,
    shared_xaxes=False,
    subplot_titles=[
        'Temperature', 'Maximum Daily Temperature', 'Minimum Daily Temperature'],
    vertical_spacing=0.12,
    horizontal_spacing=0.05
)

for field, col in zip(['tas', 'tasmax', 'tasmin'],[1, 2, 3]):
    
    # Subtract from observation
    for sim,colour,sim_name,dash_style,station_data in zip(['urban_C','rural_C','urban_T','rural_T'],
                                   ['blue','blue','red','red'],
                                   ['CLASS (urban)','CLASS (reduced urban)','TEB+CLASS (urban)','TEB+CLASS (reduced urban)'],
                                   ['solid','dot','solid','dot'],
                                   ['urban_S','rural_S', 'urban_S', 'rural_S']):
        total_fig.add_trace(go.Scatter(
            x=months,
            y=np.array(monthly[field][sim]) - np.array(monthly[field][station_data]) - np.ones(12)*273.15,
            name=f'{sim_name}',
            legendgroup=f'{sim_name} ',
            line=dict(color=colour, dash=dash_style,width=1.5),
            showlegend=(col == 1)
        ), row=1, col=col)
        total_fig.add_shape(
            type='line',
            x0=months[0], x1=months[-1],
            y0=0, y1=0,
            xref=f'x{col}', yref='y1',  
            line=dict(color='black', width=1, dash='solid'),
            layer='above'
        )

dtick = 0.5
total_fig.update_yaxes(title_text='Temperature Difference from Observation (°C)', row=1, col=1,dtick=dtick)
for col in [2, 3]:
    total_fig.update_yaxes(
        dtick=dtick,
        row=1, col=col
    )
total_fig.update_layout(
    template='plotly_white',
    legend=dict(),
)

total_fig.show()
# total_fig.write_html('/home/gulley/UHI_HW_MTL/info/plots/monthly_grouped_obsdiff_temperature_cycles.html')
```
