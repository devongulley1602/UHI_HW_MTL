```python
import xarray as xr
from Montreal_UHI_toolbox import add_field_to_stations, add_blurred_field_to_stations, stations

# Two methods for comparing stations and observation:
    # 1. Smooth JJA tasmin/max/avg simulated fields, grab closest station
    # 2. Use multivariate regression model (MVRM) trained on station data to reproduce the simulation
```

```python
# Load all seasonal data from observation and otherwise into stations dataset

seasonal = {}
seasonal_std = {}
path = '/runoff/gulley/St_Laurent/intermediates'
season = 'JJA'
for field in ['tasmin','tasmax','tasavg']:
    seasonal[f'{field}_C'] = xr.open_zarr(f'{path}/sim/seasons_avg_{field}_noTEB.zarr')[field].sel(season='JJA') - 273.15
    seasonal[f'{field}_T'] = xr.open_zarr(f'{path}/sim/seasons_avg_{field}_TEB.zarr')[field].sel(season='JJA') - 273.15
    seasonal_std[f'{field}_C'] = xr.open_zarr(f'{path}/sim/seasons_std_{field}_noTEB.zarr')[field].sel(season='JJA')
    seasonal_std[f'{field}_T'] = xr.open_zarr(f'{path}/sim/seasons_std_{field}_TEB.zarr')[field].sel(season='JJA')

    # For comparing to station averages
    field1 = field
    if field == 'tasavg':
        field1 = 'tas'
    stations = add_blurred_field_to_stations(seasonal[f'{field}_C'],stations=stations,name=f'{field1}_summer_avg_C')
    stations = add_blurred_field_to_stations(seasonal[f'{field}_T'],stations=stations,name=f'{field1}_summer_avg_T')


tasmax_S = xr.open_zarr('/runoff/gulley/St_Laurent/intermediates/station/seasons_avg_tasmax.zarr')['tasmax'].sel(season=season)
tasmin_S = xr.open_zarr('/runoff/gulley/St_Laurent/intermediates/station/seasons_avg_tasmin.zarr')['tasmin'].sel(season=season)
tasavg_S = xr.open_zarr('/runoff/gulley/St_Laurent/intermediates/station/seasons_avg_tas.zarr')['tas'].sel(season=season)
for da in [tasmax_S,tasmin_S,tasavg_S]:
    stations = stations.assign_coords({f'{da.name}_summer_avg_S':da.values})
```
